```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = [7,8,5,6,9,4,10,12,15]
```

```python
plt.boxplot(
    data,
    vert=False,
    showmeans= True
)

plt.grid(True)
```

```python

# multiple dataset 

group1 = np.random.normal(50,10,100)
group2 = np.random.normal(60,15,100)

plt.boxplot(
    [group1,group2],
    tick_labels=["group1","group2"],
    showmeans=True,
    whis=2.0
)
```

```python
#stack plots
days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]
direct = [50,60,70,80,90,100,110]
organic = [30,40,50,60,70,75,80]
social = [20,25,30,40,50,55,60]


plt.stackplot(days,direct,organic,social,labels=["direct","organic","social"])

plt.title("Marketing data of the week")
plt.xlabel("Days")
plt.ylabel("No of customers")
plt.legend()
```
