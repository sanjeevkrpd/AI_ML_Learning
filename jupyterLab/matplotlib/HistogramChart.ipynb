```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
scores = np.random.normal(70,10,100) # normal distribution
```

```python
plt.hist(
     scores,
    color = "skyblue",
    edgecolor="black",
    bins = 20
        )

plt.xlabel("scores")
plt.ylabel("no of students")
```

```python
# multiple dataset

legit_transaction = [1,2,3,4,5,6]
fraud_transaction = [3,4,5,6,3,1]

plt.hist(fraud_transaction,color="red",bins=10,label="Fraud" , edgecolor="black",alpha=0.3)
plt.hist(legit_transaction,color="green",bins=10,label="Fraud" , edgecolor="black",alpha=0.5)

plt.title("Transaction Amount Distribution - Legit vs Fraud")
plt.xlabel("Amount in $")
plt.ylabel("Freq")
plt.legend()
```
