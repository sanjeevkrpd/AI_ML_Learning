```python
# Sample data
age = [21, 23, 25, 28, 30, 32, 35, 37, 40, 42,
       45, 48, 50, 53, 55, 58, 60, 63, 65, 68]

bp = [108, 110, 112, 115, 118, 119, 122, 124, 126, 128,
      130, 133, 135, 138, 140, 142, 145, 148, 150, 152]


plt.scatter(age, bp, color='red', marker='o', s=100)

plt.title("Age VS BP")
plt.xlabel("age")
plt.ylabel("bp")
plt.show()

```

```python
import matplotlib.pyplot as plt

city = [
    "Delhi",
    "Mumbai",
    "Pune",
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Kolkata",
    "Ahmedabad",
    "Jaipur",
    "Lucknow"
]

winter_hum = [52, 62, 48, 58, 47, 72, 63, 42, 40, 57]
winter_temp = [14, 24, 19, 22, 22, 25, 20, 20, 16, 15]

summer_hum = [35, 75, 45, 50, 40, 65, 75, 32, 28, 45]
summer_temp = [39, 32, 33, 31, 36, 38, 35, 40, 41, 39]

plt.scatter(winter_temp, winter_hum, label="Winter")
plt.scatter(summer_temp, summer_hum, label="Summer")

plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.title("Winter vs Summer Temperature and Humidity")

plt.legend()
plt.grid(True)

plt.show()
```
