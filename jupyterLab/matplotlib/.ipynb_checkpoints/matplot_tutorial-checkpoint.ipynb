```python
years = [2008,2009,2010,2011,2012]

oscar_movies = [
    "The dark night",
    "The hurt Locker",
    "THe King's speech",
    "The Aritist",
    "Agro"
]

oscar_revenue = [1005,170,427,133,232]


plt.plot(years,oscar_revenue,label = "oscar movie")
plt.title("Oscar Movie Revenue in each year")
plt.xlabel("years")
plt.ylabel("Revenue (in $M) ")

plt.show()
```

```python
non_oscar_movie = [
    "slumdog Millionare",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]

non_oscar_revenue = [387,2788,829,185,275]

plt.plot(years,oscar_revenue,label = "oscar movie")
plt.plot(years,non_oscar_revenue,label = "non oscar movie")
plt.title("Oscar Movie V/S Non Oscar Movie Earning ")
plt.xlabel("years")
plt.ylabel("Revenue (in $M) ")

plt.legend()

```

```python


non_oscar_movie = [
    "slumdog Millionare",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]

non_oscar_revenue = [387,2788,829,185,275]
years = [2008,2009,2010,2011,2012]
oscar_revenue = [1005,170,427,133,232]

plt.plot(years,oscar_revenue,  "o--r", label = "oscar movie")
plt.plot(years,non_oscar_revenue, "o--g" ,label = "non oscar movie")
plt.title("Oscar Movie V/S Non Oscar Movie Earning ")
plt.xlabel("years")
plt.ylabel("Revenue (in $M) ")
plt.grid()
plt.legend()
plt.savefig("pltimage.png")
```

```python

non_oscar_revenue = [387,2788,829,185,275]
years = [2008,2009,2010,2011,2012]
oscar_revenue = [1005,170,427,133,232]



plt.bar(years,non_oscar_revenue);
plt.title("Non Oscar Revenue in M ( $ )")

plt.xlabel("Years")
plt.ylabel("Millions $")

for i in range(len(non_oscar_revenue)):
    plt.text(years[i],non_oscar_revenue[i]+20,str(non_oscar_revenue[i]) ,ha="center")
```
