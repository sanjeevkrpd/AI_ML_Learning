```python
import matplotlib.pyplot as plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 24, 23, 25, 21],  # New York
    [18, 19, 17, 20, 18],  # London
    [35, 36, 38, 37, 39],  # Delhi
    [27, 28, 26, 29, 30]   # Tokyo
]
```

```python

fig, axes = plt.subplots(2,2)
city_no = 0

for i in range(2):
    for j in range(2):
        axes[i][j].plot(days,temperatures[city_no])
        axes[i][j].set_title(cities[city_no])
        city_no = city_no + 1

fig.suptitle("Temp in cities over the week")
fig.tight_layout()
        
```

```python
axes = axes.flatten()
for i,ax in enumerate(axes):
    ax.plot(days,temperatures[i],marker='o')
    ax.set_title(cities[i])
    ax.grid=True

plt.show()
```
