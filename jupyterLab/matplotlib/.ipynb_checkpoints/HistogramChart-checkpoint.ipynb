```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
scores = np.random.normal(70,10,100) # normal distribution
```

```python
plt.hist(
     scores,
    color = "skyblue",
    edgecolor="black",
    bins = 20
        )

plt.xlabel("scores")
plt.ylabel("no of students")
```
