```python
import matplotlib.pyplot as plt

expenses = ["Salary", "TutionFee", "SchoolFee", "Travelling", "Food", "HouseRent"]
amount = [2000, 500, 1000, 200, 1500, 3000]

plt.pie(
    amount,
    labels=expenses,
    autopct="%1.1f%%",
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1,
        "linestyle": "-",
        "fill": False
    },
    shadow=True
)

plt.title("Expense Distribution")

plt.show()
```
