```python
from bs4 import BeautifulSoup
```

```python
with open("scrap_data/data1.html" , "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "html.parser")
```

```python

h3_all = soup.find_all("h3")
```

```python
country = []
population = []
capital = []
area = []
h3_all

for h3 in h3_all:
    country.append(h3.get_text(strip=True))
    population.append(h3.find_next("div").select("span.country-population")[0].get_text(strip=True))
    capital.append(h3.find_next("div").select("span.country-capital")[0].get_text(strip=True))
    area.append(h3.find_next("div").select("span.country-area")[0].get_text(strip=True))
```

```python
df = pd.DataFrame({
    "country": country,
    "population": population,
    "capital" : capital,
    "area" : area
})

print(df.head())
```

```python
df.to_csv("scrap_data/cleared_data.csv", index=False)
```
